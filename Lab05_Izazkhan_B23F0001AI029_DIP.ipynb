#**Task 01**

```python
# 1. Imports and File Upload
import cv2
import numpy as np
from google.colab import files
from google.colab.patches import cv2_imshow

uploaded = files.upload()
filename = list(uploaded.keys())[0]

# Show original image
image = cv2.imread(filename)
print("Step 1: Original Security Footage")
cv2_imshow(image)
```

```python
# 2. Convert to Grayscale and Blur
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
blurred = cv2.GaussianBlur(gray, (5, 5), 0)

# Show the pre-processed image
print("Step 2: Grayscale and Blurred (Noise Reduction)")
cv2_imshow(blurred)
```

```python
# 3. Apply Canny Edge Detection
low_edges = cv2.Canny(blurred, 50, 100)
high_edges = cv2.Canny(blurred, 150, 250)

# Show the comparison
print("Step 3: Low Threshold (Shows more detail/texture)")
cv2_imshow(low_edges)

print("Step 3: High Threshold (Shows only strong outlines)")
cv2_imshow(high_edges)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Convert edges to 3-channel image. Using high_edges for strong outlines.
edges_color = cv2.cvtColor(high_edges, cv2.COLOR_GRAY2BGR)

# Overlay edges on original image
overlay = cv2.addWeighted(image, 0.8, edges_color, 0.8, 0)

plt.imshow(cv2.cvtColor(overlay, cv2.COLOR_BGR2RGB))
plt.title("Edges Overlay on Original")
plt.axis('off')
plt.show()
```

#**Task 02**

```python
import cv2
import numpy as np
from google.colab import files
from matplotlib import pyplot as plt

# Trigger the upload dialog
uploaded = files.upload()
filename = list(uploaded.keys())[0]

# Load image in Grayscale (best for edge detection)
img = cv2.imread(filename, cv2.IMREAD_GRAYSCALE)
plt.imshow(img, cmap='gray')
plt.title("Original Road Sign")
plt.show()
```

```python
# Apply Laplacian and Sobel filters
laplacian = cv2.Laplacian(img, cv2.CV_64F, ksize=3)
sobel_x = cv2.Sobel(img, cv2.CV_64F, 1, 0, ksize=3)
sobel_y = cv2.Sobel(img, cv2.CV_64F, 0, 1, ksize=3)

# Combine Sobel results
sobel_combined = cv2.sqrt(cv2.pow(sobel_x, 2) + cv2.pow(sobel_y, 2))
# Create a plot to show the original and processed images
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1), plt.imshow(img, cmap='gray'), plt.title('Original')
plt.subplot(1, 3, 2), plt.imshow(laplacian, cmap='gray'), plt.title('Laplacian Edges')
plt.subplot(1, 3, 3), plt.imshow(sobel_combined, cmap='gray'), plt.title('Sobel Combined')

plt.tight_layout()
plt.show()
```

```python
# Applying different kernel sizes for comparison
laplacian_small = cv2.Laplacian(img, cv2.CV_64F, ksize=3)
laplacian_large = cv2.Laplacian(img, cv2.CV_64F, ksize=7)

# Displaying for the Word file
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1), plt.imshow(laplacian_small, cmap='gray'), plt.title('3x3 Kernel (Sharp Detail)')
plt.subplot(1, 2, 2), plt.imshow(laplacian_large, cmap='gray'), plt.title('7x7 Kernel (Blurred/Thick)')
plt.show()
```

#**Task 03**

```python
# Import necessary libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# Upload image from local system
uploaded = files.upload()

# Get the uploaded image name
image_path = list(uploaded.keys())[0]

# Read image in grayscale
image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)

# Display the original image
plt.figure(figsize=(6,6))
plt.imshow(image, cmap='gray')
plt.title("Original X-ray Image")
plt.axis('off')
plt.show()
```

```python
# Apply Laplacian with different kernel sizes
laplacian_1 = cv2.Laplacian(image, cv2.CV_64F, ksize=1)
laplacian_3 = cv2.Laplacian(image, cv2.CV_64F, ksize=3)
laplacian_5 = cv2.Laplacian(image, cv2.CV_64F, ksize=5)

# Convert to absolute values and uint8
laplacian_1 = np.uint8(np.absolute(laplacian_1))
laplacian_3 = np.uint8(np.absolute(laplacian_3))
laplacian_5 = np.uint8(np.absolute(laplacian_5))
```

```python
# Display all results
plt.figure(figsize=(15,5))

plt.subplot(1,4,1)
plt.imshow(image, cmap='gray')
plt.title("Original")
plt.axis('off')

plt.subplot(1,4,2)
plt.imshow(laplacian_1, cmap='gray')
plt.title("Laplacian k=1")
plt.axis('off')

plt.subplot(1,4,3)
plt.imshow(laplacian_3, cmap='gray')
plt.title("Laplacian k=3")
plt.axis('off')

plt.subplot(1,4,4)
plt.imshow(laplacian_5, cmap='gray')
plt.title("Laplacian k=5")
plt.axis('off')

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# Upload image
uploaded = files.upload()
image_path = list(uploaded.keys())[0]

# Read image
image = cv2.imread(image_path)
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

# Display original image
plt.figure(figsize=(6,6))
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.title("Original Satellite Image")
plt.axis('off')
plt.show()
```

```python
# Optional: Apply Gaussian Blur to reduce noise
blurred = cv2.GaussianBlur(gray, (5,5), 0)

# Apply Canny with different thresholds
edges_1 = cv2.Canny(blurred, 50, 150)
edges_2 = cv2.Canny(blurred, 100, 200)
edges_3 = cv2.Canny(blurred, 150, 300)
```

```python
plt.figure(figsize=(15,5))

plt.subplot(1,4,1)
plt.imshow(gray, cmap='gray')
plt.title("Grayscale")
plt.axis('off')

plt.subplot(1,4,2)
plt.imshow(edges_1, cmap='gray')
plt.title("Canny (50,150)")
plt.axis('off')

plt.subplot(1,4,3)
plt.imshow(edges_2, cmap='gray')
plt.title("Canny (100,200)")
plt.axis('off')

plt.subplot(1,4,4)
plt.imshow(edges_3, cmap='gray')
plt.title("Canny (150,300)")
plt.axis('off')

plt.show()
```
